```python
from datasets import Dataset, Features, Image, Value
import json

jsonl_path = "/bigdisk/minhpvt/quick-test/na/ML/data/batches/image_captions_1_34.jsonl"

# Load data from JSONL
data = []
with open(jsonl_path, 'r') as f:
    for line in f:
        data.append(json.loads(line))

# Assuming fields: "image_path" and "caption" (adjust if different)
features = Features({"image": Image(), "caption": Value("string")})
ds = Dataset.from_dict({
    "image": [d["image_path"] for d in data],
    "caption": [d["description"] for d in data]
})
ds = ds.cast(features)
```
